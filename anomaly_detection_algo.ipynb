In the above image, its easier  to seperate points D, E (with one cut across the plane), rather than the points B , C. B requires around 4 cuts to be completely isolated. A requires 2 cuts, and C requires around 3 cuts. 

When we run this isolation forest algorithm through the data, the outliers get a negative score, while the inliers receive a positive score.

This algo is based on the assumption that anomalies are few and different from other data, and hence they can be isolated using a few partitions.

Invariant to the choice of distance metrics.

# Principle




If an instance has a shorter path then its more likely to be an anomaly as it was easier to split from the rest of the data using this randome split proces.

The above was one isolation tree, there is a bit of randomness there, we have a chance of splitting an inlier . So that why we need a collection of trees rather than a single tree and capture the average path length of a point.

if s is the anomaly score, h(x) is the path length of x in one of those isolation trees, E[h(x)] is the average path length of instance x across all isolation trees. c(n) is normalized value, if we have larger samples then path length will be larger,
 
c(n) is the average depth of all the points(each point calculated across all decision trees).

If E(h(x))<< c(n), s ->1, its a oulier 

If E(h(x)) = c(n), s - > 0.5, normal point

If E(h(x))>> c(n) , s-> 0 (<0.5) , not a outlier

2. # **LOCAL OUTLIER FACTOR**

The above outlier techniques identify only the global outlier and not the local outliers. To identify the local outliers, we go for density based techniques.

Global identifiers are easier to find as they are isoated.Local outliers are trickier to find since they are near clusters.
This is another anomaly detection method, but instead of isolating points like Isolation Forest, LOF measures how densely a point is surrounded compared to its neighbors.



KNN is great at detecting global outliers but not the local outliers.

Density of a observation is the inverse of average reachability from the ibdervation to all of its k nearest neighbour. Less dense means we travel more to reach the nearest point of the obs.

so if average density of the knn of an observation > average density of the obs, score (A) >1 , meaning its an outlier

else if average density of the knn of an obs < avg density of the obs, score(A)<1, meaning its an inlier.
